## 4. Cleaning Actions

The identified data-quality issues were handled carefully without fabricating missing information.

| Issue | Action Taken | Result |
|---|---|---|
| Genuine missing values | Retained missing values where no reliable source value was available | 4,307 missing cells remained |
| Exact duplicate rows | No deletion required | 0 duplicates |
| Duplicate `show_id` values | No correction required | 0 duplicates |
| Extra/repeated whitespace | Trimmed and normalized whitespace | 119 formatting issues reduced to 0 |
| `date_added` data type | Converted from text to datetime | All 8,797 non-missing dates parsed successfully |
| `release_year` data type | Converted to nullable integer | No invalid values |
| Duration stored in `rating` | Moved 3 valid movie durations to `duration` | 3 misplaced values corrected |
| Duration formatting | Validated movie and TV-show duration formats | No invalid duration formats |
| Rating formatting | Removed duration-like values from `rating` | 0 duration-like values remained |
| `type` consistency | Validated categorical values | Only `Movie` and `TV Show` found |
| Date formatting | Standardized exported dates to `YYYY-MM-DD` | Consistent date format |

```python
# Load the raw dataset

INPUT = Path(r"D:\DATA ANALYTICS VIRTUAL INTERNSHIP\SWYNEX TECHNOLOGY  INTERNSHIP\netflix_titles.csv")

df = pd.read_csv(
    INPUT,
    dtype="string",
    keep_default_na=True
)

print("Raw dataset loaded successfully.")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
# Standardize whitespace in text columns

text_cols = [
    "title", "director", "cast", "country", "date_added",
    "rating", "duration", "listed_in", "description"
]

for col in text_cols:
    df[col] = (
        df[col]
        .str.replace("\u00A0", " ", regex=False)
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

print("Whitespace standardization completed.")
```

```python
# Correct duration values that were incorrectly stored in the rating column

misplaced_duration = (
    df["type"].eq("Movie")
    & df["rating"].str.fullmatch(r"\d+ min", case=False, na=False)
    & df["duration"].isna()
)

df.loc[misplaced_duration, "duration"] = df.loc[
    misplaced_duration, "rating"
]

df.loc[misplaced_duration, "rating"] = pd.NA

print("Misplaced duration values corrected:", misplaced_duration.sum())
```

```python
# Convert columns to appropriate data types

df["date_added"] = pd.to_datetime(
    df["date_added"],
    format="%B %d, %Y",
    errors="coerce"
)

df["release_year"] = pd.to_numeric(
    df["release_year"],
    errors="coerce"
).astype("Int64")

print("Data type conversion completed.")
print("date_added:", df["date_added"].dtype)
print("release_year:", df["release_year"].dtype)
```

```python
# Final validation checks

print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Exact duplicate rows:", df.duplicated().sum())
print("Duplicate show_id values:", df["show_id"].duplicated().sum())
print("Total missing cells:", df.isna().sum().sum())
print("Duration-like values still in rating:",
      df["rating"].str.fullmatch(r"\d+ min", case=False, na=False).sum())

print("\nData types:")
print(df[["date_added", "release_year"]].dtypes)
```

```python
# Export the cleaned dataset

OUTPUT = Path(
    r"D:\DATA ANALYTICS VIRTUAL INTERNSHIP\SWYNEX TECHNOLOGY  INTERNSHIP\netflix_titles_cleaned.csv"
)

df.to_csv(
    OUTPUT,
    index=False,
    date_format="%Y-%m-%d"
)

print("Cleaned dataset exported successfully.")
print("File:", OUTPUT)
```

```python
# Verify the exported cleaned dataset

cleaned_df = pd.read_csv(OUTPUT)

print("Cleaned dataset loaded successfully.")
print("Rows:", cleaned_df.shape[0])
print("Columns:", cleaned_df.shape[1])
print("Exact duplicate rows:", cleaned_df.duplicated().sum())
print("Duplicate show_id values:", cleaned_df["show_id"].duplicated().sum())
print("Total missing cells:", cleaned_df.isna().sum().sum())
```
